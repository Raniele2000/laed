```python
#2
from NoDuplo import *


def atualizar_elemento(cabeca, x, y):
    atual = cabeca
    while atual is not None and atual.dado != x:
        atual = atual.proximo

    if atual is None:
        return cabeca

    atual.dado = y


    if atual.anterior is not None:
        atual.anterior.proximo = atual.proximo
    else:
        cabeca = atual.proximo

    if atual.proximo is not None:
        atual.proximo.anterior = atual.anterior

    atual.proximo = None
    atual.anterior = None


    if cabeca is None:
        return atual

    if y <= cabeca.dado:
        atual.proximo = cabeca
        cabeca.anterior = atual
        return atual

    nav = cabeca
    while nav.proximo is not None and nav.proximo.dado < y:
        nav = nav.proximo


    atual.proximo = nav.proximo
    if nav.proximo is not None:
        nav.proximo.anterior = atual
    nav.proximo = atual
    atual.anterior = nav

    return cabeca


lista_2 = criar_lista_dupla([3, 5, 9, 10, 15])
print("2. Operação de atualização (atualizar 9 para 1)")
imprimir_lista_dupla(lista_2)
lista_2 = atualizar_elemento(lista_2, 9, 1)
imprimir_lista_dupla(lista_2)
print("-" * 30)

"""
b) Tempo de execução:
O(n). A operação realiza, no pior dos casos, uma varredura para localizar o elemento x (O(n)) e, em seguida, outra
varredura para encontrar a posição correta de inserção do valor y modificado (O(n)). O resultado líquido é O(n).
"""
```

```python
#3
from NoDuplo import *


def trocar_nos(cabeca, no1, no2):
    if no1 == no2:
        return cabeca
    if no2.proximo == no1:
        no1, no2 = no2, no1

    ant1, prox1 = no1.anterior, no1.proximo
    ant2, prox2 = no2.anterior, no2.proximo

    if prox1 == no2:
        no1.proximo = prox2
        if prox2:
            prox2.anterior = no1
        no2.anterior = ant1
        if ant1:
            ant1.proximo = no2
        no2.proximo = no1
        no1.anterior = no2
    else:
        no1.proximo = prox2
        if prox2:
            prox2.anterior = no1
        no1.anterior = ant2
        if ant2:
            ant2.proximo = no1
        no2.proximo = prox1
        if prox1:
            prox1.anterior = no2
        no2.anterior = ant1
        if ant1:
            ant1.proximo = no2

    if cabeca == no1:
        cabeca = no2
    elif cabeca == no2:
        cabeca = no1

    return cabeca


def varredura(cabeca):
    if cabeca is None:
        return None

    atual = cabeca
    while atual.proximo is not None:
        if atual.dado > atual.proximo.dado:
            proximo_no = atual.proximo
            cabeca = trocar_nos(cabeca, atual, proximo_no)

        else:
            atual = atual.proximo
    return cabeca


lista_3 = criar_lista_dupla([9, 3, 8, 5, 1])
print("3. Varredura (uma passada)")
imprimir_lista_dupla(lista_3)
lista_3 = varredura(lista_3)
imprimir_lista_dupla(lista_3)
print("-" * 30)

"""
b) Tempo de execução:
O(n). A operação de varredura (uma passada do algoritmo tipo Bubble Sort) exige percorrer a lista inteira do início ao
fim uma única vez. Como a função de trocar os nós faz apenas ajustes de ponteiros (tempo O(1)), o tempo de execução é
linear em relação ao número de elementos.
"""
```

```python
#4
from NoDuplo import *


def particionar_lista(cabeca, k):
    if not cabeca or not cabeca.proximo:
        return cabeca

    q = cabeca
    r = cabeca
    while r.proximo is not None:
        r = r.proximo

    while q != r and q.anterior != r:
        while q != r and q.anterior != r and q.dado <= k:
            q = q.proximo

        while q != r and q.anterior != r and r.dado > k:
            r = r.anterior

        if q != r and q.anterior != r:
            cabeca = trocar_nos(cabeca, q, r)
            q, r = r, q
            q = q.proximo
            r = r.anterior

    return cabeca


lista_4 = criar_lista_dupla([12, 5, 9, 17, 13])
print("4. Partição (k=10)")
imprimir_lista_dupla(lista_4)
lista_4 = particionar_lista(lista_4, 10)
imprimir_lista_dupla(lista_4)
print("-" * 30)

"""
b) Tempo de execução:
O(n). Os dois ponteiros (q e r) começam nas extremidades e se movem um em direção ao outro. Cada conexão da lista é
verificada exatamente uma vez, e as trocas requerem tempo constante. A complexidade é rigorosamente linear.
"""
```

```python
#5
from NoDuplo import *


def construir_vetor_esparso(vetor):
    ficticio = NoDuplo(0)
    cauda = ficticio

    for i in range(len(vetor)):
        if vetor[i] != 0:
            novo_no = NoDuplo(vetor[i], i)
            cauda.proximo = novo_no
            novo_no.anterior = cauda
            cauda = novo_no
    cabeca_real = ficticio.proximo
    if cabeca_real:
        cabeca_real.anterior = None

    return cabeca_real


V = [0, 3, 0, 0, 0, 5, 0, 2, 0, 0, 8, 0, 0, 7, 0]
print("5. Vetor Esparso (criação)")
print("Entrada:", V)
lista_esparsa = construir_vetor_esparso(V)
imprimir_lista_dupla(lista_esparsa)
print("-" * 30)

"""
b) Tempo de execução:
O(N). O algoritmo examina sequencialmente cada um dos elementos (N) do array original para testar a validade
(diferente de zero). Mesmo gerando uma estrutura compacta, a leitura original dita um desempenho dependente do tamanho
real do array (O(N)).
"""
```

```python
#6
from NoDuplo import


def busca_por_indice(cabeca, k):
    atual = cabeca
    while atual is not None:
        if atual.indice == k:
            return atual.dado
        if atual.indice > k:
            break
        atual = atual.proximo
    return 0


def busca_por_valor(cabeca, x):  # Retorna índice do valor x
    atual = cabeca
    while atual is not None:  # Percorre todos os registros
        if atual.dado == x:  # Compara os valores
            return atual.indice  # Retorna onde ele existia
        atual = atual.proximo
    return -1  # Se não encontrou o valor, retorna -1


def atualizacao_esparso(cabeca, x, k):  # Atualiza o índice k para o valor x
    ficticio = NoDuplo(0)  # Nó fantasma para evitar vazamentos na cabeça
    ficticio.proximo = cabeca
    if cabeca:
        cabeca.anterior = ficticio

    atual = cabeca
    anterior = ficticio  # Mantemos rastro do vizinho

    while atual is not None and atual.indice < k:  # Navega até o índice desejado
        anterior = atual
        atual = atual.proximo

    if atual is not None and atual.indice == k:  # O índice já tem um nó existente
        if x == 0:  # Quer atualizar para zero, logo devemos deletar esse nó
            anterior.proximo = atual.proximo
            if atual.proximo:
                atual.proximo.anterior = anterior
        else:  # Caso seja diferente de zero, basta sobrescrever
            atual.dado = x
    elif x != 0:  # O índice não existe (era zero) e estamos adicionando valor
        novo_no = NoDuplo(x, k)  # Cria o novo membro
        novo_no.proximo = atual  # Engata o direito
        novo_no.anterior = anterior  # Engata o esquerdo
        anterior.proximo = novo_no  # O anterior abraça ele
        if atual:
            atual.anterior = novo_no  # O sucessor o reconhece

    nova_cabeca = ficticio.proximo  # Pega de volta o limiar real
    if nova_cabeca:
        nova_cabeca.anterior = None
    return nova_cabeca


def construir_vetor_esparso(vetor):
    ficticio = NoDuplo(0)
    cauda = ficticio
    for i in range(len(vetor)):
        if vetor[i] != 0:
            novo_no = NoDuplo(vetor[i], i)
            cauda.proximo = novo_no
            novo_no.anterior = cauda
            cauda = novo_no
    cabeca_real = ficticio.proximo
    if cabeca_real:
        cabeca_real.anterior = None
    return cabeca_real


V = [0, 3, 0, 0, 0, 5, 0, 2, 0, 0, 8, 0, 0, 7, 0]
lista_esparsa = construir_vetor_esparso(V)

print("6. Operações em vetores esparso")
print("Busca-por-índice(13):", busca_por_indice(lista_esparsa, 13))  # Deve retornar 7
print("Busca-por-valor(5):", busca_por_valor(lista_esparsa, 5))  # Deve retornar Índice 5
lista_esparsa = atualizacao_esparso(lista_esparsa, 10, 8)  # Muda o 0 no índice 8 para 10
print("Após atualização (10, 8): ", end="")
imprimir_lista_dupla(lista_esparsa)
print("-" * 30)

"""
b) Tempo de execução:
Todos esses algoritmos apresentam complexidade de tempo O(E), onde E é o número de elementos não nulos armazenados na
lista esparsa. Para as três funções, percorremos a lista duplamente encadeada uma vez, o que independe do tamanho real
do vetor (zeros foram omitidos), garantindo ganho substancial de memória e busca.
"""
```

```python
#9from NoDuplo import *
import math


def transformar_em_lista_de_listas(cabeca, k):  # Recebe a cabeça original e quantos sub-vetores k gerar
    if cabeca is None or k <= 0:
        return []  # Prevenção

    # Passo 1: Conta o número total de elementos n
    n = 0
    atual = cabeca
    while atual is not None:
        n += 1
        atual = atual.proximo

    # Passo 2: Define o tamanho aproximado das sublistas
    tamanho_sub = math.ceil(n / k)  # Usa o teto para não deixar elementos perdidos

    lista_de_listas = []  # O vetor em Python que servirá de contêiner de ponteiros
    atual = cabeca

    # Passo 3: Quebra as sublistas e popula o array
    for _ in range(k):
        if atual is None:
            break  # Fuga de emergência se os cálculos acabarem antes

        lista_de_listas.append(atual)  # Adiciona a cabeça da sublista no array

        contador = 1
        while contador < tamanho_sub and atual.proximo is not None:  # Avança ao fim dessa fatia
            atual = atual.proximo
            contador += 1

        proximo_bloco = atual.proximo  # Salva o líder da próxima fatia
        atual.proximo = None  # Rompe a conexão com a próxima sublista (tesoura)
        if proximo_bloco is not None:
            proximo_bloco.anterior = None  # Limpa o referencial da nova cabeça gerada

        atual = proximo_bloco  # Transita o cursor geral para preparar o loop superior

    return lista_de_listas  # Retorna o array de sublistas


lista_9 = criar_lista_dupla([1, 3, 7, 10, 13, 18, 21, 27])
print("9. Lista de listas (k=3)")
L = transformar_em_lista_de_listas(lista_9, 3)
for idx, sublista in enumerate(L):
    print(f"Ponteiro [{idx}]: ", end="")
    imprimir_lista_dupla(sublista)
print("-" * 30)

"""
b) Tempo de execução:
O(n). A conversão exige uma contagem global para derivar os tamanhos paritários O(n) e, em sequência, um laço de
repetição que fragmenta a cadeia nos limites apropriados (também O(n)). Como não há manipulação sobreposta de ciclos,
a conversão toda finaliza linearmente em O(n).
"""
```

```python
#10
from NoDuplo import *
import math


def busca_lista_de_listas(L, x):  # Recebe o array de ponteiros e o valor a se buscar
    # Busca binária superficial para descobrir de qual sublista este número pode pertencer
    inicio, fim = 0, len(L) - 1
    indice_alvo = -1

    while inicio <= fim:  # Restringe a busca baseado na cabeça das sublistas
        meio = (inicio + fim) // 2
        if L[meio] is None:
            break

        if L[meio].dado == x:  # Se calhar de ser o líder da sublista
            return True
        elif L[meio].dado < x:  # Se o líder for menor, o número pode estar ali ou mais pra direita
            indice_alvo = meio
            inicio = meio + 1
        else:  # Se o líder é maior, então obrigatoriramente está nas listas anteriores
            fim = meio - 1

    if indice_alvo == -1:
        return False  # Se for menor que todo mundo, não tá na lista

    # Travessia apenas na sublista designada
    atual = L[indice_alvo]
    while atual is not None:
        if atual.dado == x:
            return True
        if atual.dado > x:  # Aborta precocemente se as grandezas excederem o buscado
            break
        atual = atual.proximo

    return False


def insercao_lista_de_listas(L, x):  # Recebe array e valor a ser criado
    if not L or L[0] is None:
        return  # Se vazio

    # Determina quem assume a adoção do elemento com lógica linear simples por simplificação
    indice_alvo = 0
    for i in range(1, len(L)):
        if L[i] is not None and L[i].dado <= x:
            indice_alvo = i
        else:
            break

    novo_no = NoDuplo(x)  # Gera o bloco de memória
    atual = L[indice_alvo]

    if x <= atual.dado:  # Inserção no começo afeta a representação local
        novo_no.proximo = atual
        atual.anterior = novo_no
        L[indice_alvo] = novo_no  # O array de ponteiros é atualizado
        return

    while atual.proximo is not None and atual.proximo.dado < x:  # Navega a sublista
        atual = atual.proximo

    # Costura o elemento no seu nicho ordenado
    novo_no.proximo = atual.proximo
    if atual.proximo:
        atual.proximo.anterior = novo_no
    atual.proximo = novo_no
    novo_no.anterior = atual


def remocao_lista_de_listas(L, x):  # Deleta elementos se existirem
    for i in range(len(L)):
        atual = L[i]
        # Pula as listas em que não é possível constar
        if atual is None or (i < len(L) - 1 and L[i + 1] is not None and x >= L[i + 1].dado):
            continue

        while atual is not None:
            if atual.dado == x:  # Fizeram "Match" e devemos extraí-lo
                if atual.anterior:
                    atual.anterior.proximo = atual.proximo
                else:  # Caso onde estarmos removendo a cabeça guardada no Array
                    L[i] = atual.proximo
                if atual.proximo:
                    atual.proximo.anterior = atual.anterior
                return True  # Termina com sucesso
            if atual.dado > x:
                break  # Fuga segura
            atual = atual.proximo
    return False


def transformar_em_lista_de_listas(cabeca, k):
    if cabeca is None or k <= 0:
        return []
    n = 0
    atual = cabeca
    while atual is not None:
        n += 1
        atual = atual.proximo
    tamanho_sub = math.ceil(n / k)
    lista_de_listas = []
    atual = cabeca
    for _ in range(k):
        if atual is None:
            break
        lista_de_listas.append(atual)
        contador = 1
        while contador < tamanho_sub and atual.proximo is not None:
            atual = atual.proximo
            contador += 1
        proximo_bloco = atual.proximo
        atual.proximo = None
        if proximo_bloco is not None:
            proximo_bloco.anterior = None
        atual = proximo_bloco
    return lista_de_listas


lista_10_base = criar_lista_dupla([1, 3, 7, 10, 13, 18, 21, 27])
L = transformar_em_lista_de_listas(lista_10_base, 3)

print("10. Operações na lista de listas")
print("Busca por 18:", busca_lista_de_listas(L, 18))
print("Inserindo o valor 15...")
insercao_lista_de_listas(L, 15)
for idx, sublista in enumerate(L):
    print(f"Ponteiro [{idx}]: ", end="")
    imprimir_lista_dupla(sublista)
print("Removendo o valor 7...")
remocao_lista_de_listas(L, 7)
for idx, sublista in enumerate(L):
    print(f"Ponteiro [{idx}]: ", end="")
    imprimir_lista_dupla(sublista)

"""
b) Tempo de execução:
Todas as operações tiram extrema vantagem da fragmentação (se o total é n, cada fragmento guarda n/k nós).
Busca: O(log k + n/k), pois utilizamos busca binária sobre os k ponteiros primários e busca exata local.
Inserção/Remoção: O(k + n/k) caso utilizem iteração inicial pelo array, ou até O(log k + n/k) se otimizadas. Reduzir a
 lista principal diminui severamente o atrito e custo constante exigido pelos algoritmos padrões.
"""
```

```python
# no duplo
class NoDuplo:  # Classe base para os elementos da lista duplamente encadeada
    def __init__(self, dado, indice=None):  # Construtor do nó
        self.dado = dado  # Armazena o valor do nó
        self.indice = indice  # Armazena o índice (útil para vetores esparsos)
        self.anterior = None  # Ponteiro para o nó anterior
        self.proximo = None  # Ponteiro para o nó seguinte


def criar_lista_dupla(valores):  # Função auxiliar para criar lista para os testes
    if not valores:
        return None
    cabeca = NoDuplo(valores[0])
    atual = cabeca
    for val in valores[1:]:
        novo_no = NoDuplo(val)
        atual.proximo = novo_no
        novo_no.anterior = atual
        atual = novo_no
    return cabeca


def imprimir_lista_dupla(cabeca):  # Função auxiliar para imprimir visualmente a lista
    valores = []
    atual = cabeca
    while atual is not None:
        if atual.indice is not None:
            valores.append(f"[{atual.dado}|i:{atual.indice}]")
        else:
            valores.append(str(atual.dado))
        atual = atual.proximo
    print(" <-> ".join(valores))


def trocar_nos(cabeca, no1, no2):  # Função auxiliar isolada para trocar nós
    if no1 == no2:
        return cabeca  # Não faz nada se forem o mesmo nó

    if no2.proximo == no1:  # Garante que no1 vem antes de no2 logicamente
        no1, no2 = no2, no1

    ant1, prox1 = no1.anterior, no1.proximo  # Salva conexões do no1
    ant2, prox2 = no2.anterior, no2.proximo  # Salva conexões do no2

    if prox1 == no2:  # Caso sejam adjacentes
        no1.proximo = prox2
        if prox2:
            prox2.anterior = no1
        no2.anterior = ant1
        if ant1:
            ant1.proximo = no2
        no2.proximo = no1
        no1.anterior = no2
    else:  # Caso sejam nós separados
        no1.proximo = prox2
        if prox2:
            prox2.anterior = no1
        no1.anterior = ant2
        if ant2:
            ant2.proximo = no1
        no2.proximo = prox1
        if prox1:
            prox1.anterior = no2
        no2.anterior = ant1
        if ant1:
            ant1.proximo = no2

    if cabeca == no1:  # Arruma a cabeça se necessário
        cabeca = no2
    elif cabeca == no2:
        cabeca = no1

    return cabeca  # Retorna a cabeça correta
```
