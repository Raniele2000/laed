```python
#1
from No import *


def mover_maior_para_fim(cabeca):
    if cabeca is None or cabeca.proximo is None:
        return cabeca

    valor_max = cabeca.dado
    anterior_max = None
    no_max = cabeca
    anterior = cabeca
    atual = cabeca.proximo
    while atual is not None:
        if atual.dado > valor_max:
            valor_max = atual.dado
            anterior_max = anterior
            no_max = atual
        anterior = atual
        atual = atual.proximo

    if no_max.proximo is None:
        return cabeca

    if anterior_max is None:
        cabeca = cabeca.proximo
    else:
        anterior_max.proximo = no_max.proximo

    cauda = cabeca
    while cauda.proximo is not None:
        cauda = cauda.proximo

    cauda.proximo = no_max
    no_max.proximo = None

    return cabeca


lista_1 = criar_lista([5, 8, 13, 2, 10])

print("1. O maior no fim")
print("Entrada: ", end="")
imprimir_lista(lista_1)

lista_1 = mover_maior_para_fim(lista_1)
print("Saída:   ", end="")
imprimir_lista(lista_1)
print("-" * 30)

"""
Tempo de execução: O(n). O algoritmo percorre a lista inteira duas vezes (uma para achar o maior, outra para achar a
cauda), o que configura uma complexidade linear.
"""
```

```python
#2
from No import *


def separar_pares_impares(cabeca):
    pares_cabeca = None
    pares_cauda = None
    impares_cabeca = None
    impares_cauda = None
    atual = cabeca

    while atual is not None:
        if atual.dado % 2 == 0:
            if pares_cabeca is None:
                pares_cabeca = atual
                pares_cauda = pares_cabeca
            else:
                pares_cauda.proximo = atual
                pares_cauda = atual
        else:
            if impares_cabeca is None:
                impares_cabeca = atual
                impares_cauda = impares_cabeca
            else:
                impares_cauda.proximo = atual
                impares_cauda = atual

        atual = atual.proximo
    if pares_cauda is not None:
        pares_cauda.proximo = None
    if impares_cauda is not None:
        impares_cauda.proximo = None

    return pares_cabeca, impares_cabeca

lista_2 = criar_lista([2, 8, 5, 10, 7])

print("2. Separando pares e ímpares")
print("Entrada original: ", end="")
imprimir_lista(lista_2)

pares, impares = separar_pares_impares(lista_2)
print("Lista de Pares:   ", end="")
imprimir_lista(pares)
print("Lista de Ímpares: ", end="")
imprimir_lista(impares)
print("-" * 30)

"""
Tempo de execução: O(n). O algoritmo necessita atravessar a lista vinculada de n elementos exatamente uma vez para
distribuir todos os itens.
"""
```

```python
#3
from No import *


def inverter_lista(cabeca):
    anterior = None
    atual = cabeca

    while atual is not None:
        proximo_no = atual.proximo
        atual.proximo = anterior
        anterior = atual
        atual = proximo_no

    return anterior


lista_3 = criar_lista([3, 2, 5, 9, 4])

print("3. Virando ao contrário")
print("Entrada: ", end="")
imprimir_lista(lista_3)

lista_3 = inverter_lista(lista_3)
print("Saída:   ", end="")
imprimir_lista(lista_3)
print("-" * 30)

"""
Tempo de execução: O(n). Apenas um laço visita cada nó na lista uma única vez e as operações internas requerem tempo
constante O(1).
"""
```

```python
#4
from No import *


def duplicar_impares(cabeca):
    atual = cabeca

    while atual is not None:
        if atual.dado % 2 != 0:
            novo_no = No(atual.dado)
            novo_no.proximo = atual.proximo
            atual.proximo = novo_no
            atual = novo_no.proximo
        else:
            atual = atual.proximo

    return cabeca


lista_4 = criar_lista([2, 7, 6, 3])

print("4. Duplicando ímpares")
print("Entrada: ", end="")
imprimir_lista(lista_4)

lista_4 = duplicar_impares(lista_4)
print("Saída:   ", end="")
imprimir_lista(lista_4)
print("-" * 30)

"""
Tempo de execução: O(n). Todo nó é checado somente uma vez e as eventuais adições não exigem percurso adicional pela
lista encadeada.
"""
```

```python
#5
from No import *


def intercalar_listas(p1, p2):
    ficticio = No(0)
    cauda = ficticio

    while p1 is not None and p2 is not None:
        if p1.dado <= p2.dado:
            cauda.proximo = p1
            p1 = p1.proximo
        else:
            cauda.proximo = p2
            p2 = p2.proximo

        cauda = cauda.proximo
    if p1 is not None:
        cauda.proximo = p1
    if p2 is not None:
        cauda.proximo = p2

    return ficticio.proximo


p1 = criar_lista([3, 6, 7, 10, 13])
p2 = criar_lista([2, 4, 9, 11, 12])

print("5. Intercalação")
print("Lista 1: ", end="")
imprimir_lista(p1)
print("Lista 2: ", end="")
imprimir_lista(p2)

lista_intercalada = intercalar_listas(p1, p2)
print("Fundida: ", end="")
imprimir_lista(lista_intercalada)
print("-" * 30)

"""
Tempo de execução: O(n + m). Onde n é o número de elementos em p1 e m o número de elementos em p2. O laço percorre as
listas conjuntas passo-a-passo.
"""

```

```python
#6
from No import *


def particionar_lista(cabeca, k):
    menor_cabeca = None
    menor_cauda = None
    maior_cabeca = None
    maior_cauda = None

    atual = cabeca
    while atual is not None:
        if atual.dado <= k:
            if menor_cabeca is None:
                menor_cabeca = atual
                menor_cauda = menor_cabeca
            else:
                menor_cauda.proximo = atual
                menor_cauda = atual
        else:
            if maior_cabeca is None:
                maior_cabeca = atual
                maior_cauda = maior_cabeca
            else:
                maior_cauda.proximo = atual
                maior_cauda = atual

        atual = atual.proximo

    if maior_cauda is not None:
        maior_cauda.proximo = None

    if menor_cabeca is None:
        return maior_cabeca

    menor_cauda.proximo = maior_cabeca

    return menor_cabeca


lista_6 = criar_lista([9, 2, 5, 6, 1])
k = 5

print("6. Partição")
print(f"Entrada (k={k}): ", end="")
imprimir_lista(lista_6)

lista_6 = particionar_lista(lista_6, k)
print("Saída:          ", end="")
imprimir_lista(lista_6)
print("-" * 30)

"""
Tempo de execução: O(n). Embora manipule 4 ponteiros paralelos, todo processo exige unicamente uma varredura do começo
ao fim sem repetições aninhadas.
"""
```

```python
#7
from No import *


def remover_copias(cabeca, k):
    ficticio = No(0)
    ficticio.proximo = cabeca

    anterior = ficticio
    atual = cabeca

    while atual is not None:
        if atual.dado == k:
            anterior.proximo = atual.proximo
        else:
            anterior = atual

        atual = atual.proximo

    return ficticio.proximo


lista_7 = criar_lista([1, 3, 3, 2, 3, 2])
k = 3

print("7. Removendo todas as cópias")
print(f"Entrada (k={k}): ", end="")
imprimir_lista(lista_7)

lista_7 = remover_copias(lista_7, k)
print("Saída:          ", end="")
imprimir_lista(lista_7)
print("-" * 30)

"""
Tempo de execução: O(n). Para limpar as cópias o programa tem de necessariamente passar individualmente o critério
valor em todos elementos contidos no caminho encadeado.
"""
```

```python
#8
from No import *


def elemento_mais_frequente(cabeca):
    if cabeca is None:
        return None

    contagens = {}
    atual = cabeca

    while atual is not None:
        if atual.dado in contagens:
            contagens[atual.dado] += 1
        else:
            contagens[atual.dado] = 1
        atual = atual.proximo

    mais_frequente = None
    max_ocorrencias = 0

    for elemento, contagem in contagens.items():
        if contagem > max_ocorrencias:
            max_ocorrencias = contagem
            mais_frequente = elemento

    return mais_frequente


lista_8 = criar_lista([8, 3, 8, 5, 8, 3])

print("8. O elemento que aparece mais vezes")
print("Entrada: ", end="")
imprimir_lista(lista_8)

mais_frequente = elemento_mais_frequente(lista_8)
print(f"Saída:   O elemento que mais aparece é o {mais_frequente}")
print("-" * 30)

"""
Tempo de execução: O(n). As inserções e atualizações num dicionário em Python custam rotineiramente O(1). Como passa os
n itens apenas para ler, gera custo restrito à via linear O(n).
"""
```

```python
#9
from No import


def possui_repetidos(cabeca):
    vistos = set()
    atual = cabeca

    while atual is not None:
        if atual.dado in vistos:
            return True
        vistos.add(atual.dado)
        atual = atual.proximo

    return False


lista_9 = criar_lista([2, 9, 7, 4, 1])

print("9. Elemento repetido")
print("Entrada: ", end="")
imprimir_lista(lista_9)

tem_repetido = possui_repetidos(lista_9)
print(f"Saída:   A lista possui elementos repetidos? {'Sim' if tem_repetido else 'Não'}")
print("-" * 30)

"""
Tempo de execução: O(n). Consultar ou preencher elementos num Conjunto (set) do Python custa O(1) de tempo na média,
implicando que validar n passos preserva O(n) da travessia.
"""
```

```python
#10
from No import


def intersecao_listas(p1, p2):
    elementos_p1 = set()
    atual = p1
    while atual is not None:
        elementos_p1.add(atual.dado)
        atual = atual.proximo

    ficticio = No(0)
    cauda = ficticio
    atual = p2

    while atual is not None:
        if atual.dado in elementos_p1:
            cauda.proximo = No(atual.dado)
            cauda = cauda.proximo
            elementos_p1.remove(atual.dado)
        atual = atual.proximo

    return ficticio.proximo  # Entrega bloco completo ignorando o nó zero


lista_10_p1 = criar_lista([3, 9, 2, 6, 4])
lista_10_p2 = criar_lista([4, 5, 2, 9, 3])

print("10. Interseção")
print("Lista 1: ", end="")
imprimir_lista(lista_10_p1)
print("Lista 2: ", end="")
imprimir_lista(lista_10_p2)

lista_intersecao = intersecao_listas(lista_10_p1, lista_10_p2)
print("Saída:   ", end="")
imprimir_lista(lista_intersecao)
print("-" * 30)


"""
Tempo de execução: O(n + m). As operações no conjunto são efetivadas com média constante O(1). Dessa forma o tempo
geral consiste em passar pelas duas listas em sequência linear O(n) + O(m).
"""
```

```python
class No:
    def __init__(self, dado):
        self.dado = dado
        self.proximo = None


def criar_lista(valores):
    if not valores:
        return None
    cabeca = No(valores[0])
    atual = cabeca
    for val in valores[1:]:
        atual.proximo = No(val)
        atual = atual.proximo
    return cabeca


def imprimir_lista(cabeca):
    valores = []
    atual = cabeca
    while atual is not None:
        valores.append(str(atual.dado))
        atual = atual.proximo
    print(" -> ".join(valores) + " -> None")
```
