```python
class No:
    def __init__(self, dado):
        self.dado = dado
        self.proximo = None


class NoMin:
    def __init__(self, dado, minimo):
        self.dado = dado
        self.minimo = minimo
        self.proximo = None


# =====================================================================
# 1. OPERAÇÕES BÁSICAS DE PILHA ENCADEADA
# =====================================================================

class Pilha:
    def __init__(self):
        self.topo = None

    def is_empty(self):
        return self.topo is None

    def push(self, valor):
        novo_no = No(valor)
        novo_no.proximo = self.topo
        self.topo = novo_no

    def pop(self):
        if self.is_empty():
            return None
        no_removido = self.topo
        self.topo = self.topo.proximo
        valor = no_removido.dado
        no_removido.proximo = None
        return valor

    def peek(self):
        if self.is_empty():
            return None
        return self.topo.dado

    def imprimir(self):
        atual = self.topo
        valores = []
        while atual is not None:
            valores.append(str(atual.dado))
            atual = atual.proximo
        print("Topo -> " + " -> ".join(valores) + " -> None")  # Imprime formatado


# =====================================================================
# 2. FILA ENCADEADA COM PONTEIROS INICIO E FIM
# =====================================================================

class Fila:
    def __init__(self):
        self.inicio = None
        self.fim = None

    def is_empty(self):
        return self.inicio is None

    def enqueue(self, valor):
        novo_no = No(valor)
        if self.is_empty():
            self.inicio = novo_no
            self.fim = novo_no
        else:  # Caso já existam elementos
            self.fim.proximo = novo_no
            self.fim = novo_no

    def dequeue(self):
        if self.is_empty():
            return None
        no_removido = self.inicio
        self.inicio = self.inicio.proximo
        if self.inicio is None:
            self.fim = None
        valor = no_removido.dado
        no_removido.proximo = None
        return valor

    def imprimir(self):
        atual = self.inicio
        valores = []
        while atual is not None:
            valores.append(str(atual.dado))
            atual = atual.proximo
        print("Inicio -> " + " -> ".join(valores) + " -> Fim")


# =====================================================================
# 3. INVERSÃO DE FILA COM PILHA AUXILIAR
# =====================================================================

def inverter_fila(fila):
    pilha_aux = Pilha()

    while not fila.is_empty():
        valor = fila.dequeue()
        pilha_aux.push(valor)

    while not pilha_aux.is_empty():
        valor = pilha_aux.pop()
        fila.enqueue(valor)


# =====================================================================
# 4. PILHA COM MÍNIMO EM O(1)
# =====================================================================

class PilhaMin:
    def __init__(self):
        self.topo = None
    def is_empty(self):
        return self.topo is None

    def push(self, valor):
        if self.is_empty():
            min_atual = valor
        else:  # Caso já existam nós na pilha
            min_anterior = self.topo.minimo
            min_atual = min(valor, min_anterior)

        novo_no = NoMin(valor, min_atual)
        novo_no.proximo = self.topo
        self.topo = novo_no

    def pop(self):
        if self.is_empty():
            return None
        valor = self.topo.dado
        self.topo = self.topo.proximo
        return valor

    def min(self):
        if self.is_empty():
            return None
        return self.topo.minimo

    def imprimir(self):
        atual = self.topo
        valores = []
        while atual is not None:
            valores.append(f"[{atual.dado}|Min:{atual.minimo}]")
            atual = atual.proximo
        print("Topo -> " + " -> ".join(valores) + " -> None")


# =====================================================================
# 5. BALANCEAMENTO DE DELIMITADORES
# =====================================================================

def verifica_delimitadores(expressao):
    pilha = Pilha()
    pares = {')': '(', ']': '[', '}': '{'}
    for i, char in enumerate(expressao):
        if char in "([{":
            pilha.push(char)
        elif char in ")]}":
            if pilha.is_empty():
                return False, f"Erro na pos {i}: '{char}' não tem abridor."

            topo = pilha.pop()
            if topo != pares[char]:
                return False, f"Erro na pos {i}: '{char}' não fecha com '{topo}'."

    if not pilha.is_empty():
        return False, "Erro: Expressão terminou aberta (faltou fechador)."

    return True, "Expressão Válida!"


# =====================================================================
# 6. FILA COM DUAS PILHAS ENCADEADAS
# =====================================================================

class FilaComPilhas:
    def __init__(self):
        self.pilha1_entrada = Pilha()
        self.pilha2_saida = Pilha()

    def enqueue(self, valor):
        self.pilha1_entrada.push(valor)

    def dequeue(self):
        if self.pilha2_saida.is_empty():
            while not self.pilha1_entrada.is_empty():
                valor = self.pilha1_entrada.pop()
                self.pilha2_saida.push(valor)

        if self.pilha2_saida.is_empty():
            return None

        return self.pilha2_saida.pop()


# =====================================================================
# 7. RADIX SORT ESTÁVEL (USANDO NOSSA CLASSE FILA)
# =====================================================================

def radix_sort(numeros):
    if not numeros:
        return []
    max_num = max(numeros)
    digitos = len(str(max_num))

    filas = [Fila() for _ in range(10)]
    array_atual = numeros[:]
    for posicao in range(digitos):
        divisor = 10 ** posicao

        for num in array_atual:
            digito_extraido = (num // divisor) % 10
            filas[digito_extraido].enqueue(num)

        indice = 0
        for i in range(10):
            while not filas[i].is_empty():
                array_atual[indice] = filas[i].dequeue()
                indice += 1  # Avança o cursor

        print(f"Passagem {(posicao + 1)} (casa {divisor}):", array_atual)

    return array_atual


# =====================================================================
# ÁREA DE TESTES E EXEMPLOS DE EXECUÇÃO
# =====================================================================
if __name__ == "__main__":
    print("=== Q1. Operações básicas de pilha ===")
    p1 = Pilha()
    for v in [5, 17, 42]:
        p1.push(v)
    p1.push(99)
    p1.push(3)
    p1.imprimir()
    print("Pop:", p1.pop())
    print("Pop:", p1.pop())
    p1.imprimir()

    print("\n=== Q2. Fila encadeada ===")
    f1 = Fila()
    for v in [8, 15, 23]:
        f1.enqueue(v)
    f1.enqueue(7)
    f1.enqueue(11)
    f1.imprimir()
    print("Dequeue:", f1.dequeue())
    print("Dequeue:", f1.dequeue())
    f1.imprimir()

    print("\n=== Q3. Inversão de fila ===")
    f3 = Fila()
    for v in [1, 2, 3, 4]:
        f3.enqueue(v)
    print("Antes: ", end="")
    f3.imprimir()
    inverter_fila(f3)
    print("Depois: ", end="")
    f3.imprimir()

    print("\n=== Q4. Pilha com Mínimo ===")
    p4 = PilhaMin()
    p4.push(5)
    p4.push(3)
    p4.push(7)
    p4.push(1)
    p4.imprimir()
    print("Mínimo atual:", p4.min())
    print("Pop:", p4.pop())
    p4.imprimir()
    print("Mínimo atual:", p4.min())

    print("\n=== Q5. Balanceamento ===")
    testes = ["({[]})", "({[)}]", "({[]}[()]{})"]
    for t in testes:
        valido, msg = verifica_delimitadores(t)
        print(f"{t} -> {msg}")

    print("\n=== Q6. Fila com 2 Pilhas ===")
    f6 = FilaComPilhas()
    f6.enqueue(10)
    f6.enqueue(20)
    f6.enqueue(30)
    print("Dequeue:", f6.dequeue())
    f6.enqueue(40)
    print("Dequeue:", f6.dequeue())
    print("\n=== Q7. Radix Sort ===")
    # Sequência extraída fielmente do PDF
    seq = [481, 329, 143, 612, 937, 480, 256]
    radix_sort(seq)
```
